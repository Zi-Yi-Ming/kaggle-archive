```python
import subprocess, sys, os
os.chdir("/kaggle/working")
cmds = [
    [sys.executable,'-u','_pm_wheat_match.py','--matchup','A','--n','100','--seed0','1'],
    [sys.executable,'-u','_pm_wheat_match.py','--matchup','B','--level','mild','--n','100','--seed0','1'],
    [sys.executable,'-u','_pm_wheat_match.py','--matchup','B','--level','medium','--n','100','--seed0','1'],
    [sys.executable,'-u','_pm_wheat_match.py','--matchup','C','--level','mild','--n','100','--seed0','1'],
]
for c in cmds:
    print("RUN", c, flush=True)
    r = subprocess.run(c, capture_output=True, text=True)
    print((r.stdout or "")[-400:], flush=True)
    if r.stderr: print("ERR", (r.stderr or "")[-400:], flush=True)
print("ALL DONE", flush=True)
```
